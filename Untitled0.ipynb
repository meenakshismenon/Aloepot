```python
import re
text=""" Contact us at meenakshi@gmail.com .For queries mail at queries@college.edu """
email=re.findall(r'[A-za-z0-0._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}',text)
print(email)
```

```python
import re

text = "The product started in 1998.paused in 2003, and was completed in 2021."

result = re.findall(r'\d{4}', text)

print(result)
```

```python
import re

sample_text = "The cat sat on the car, but the cow chased a cap. Coding is cool!"


matches = re.findall(r'\b[cC][a-zA-Z]{2}\b', sample_text)

print("Matching words:", matches)
```

```python
import re

price_text = "The book costs $10, the jacket is $150, and the coffee is only $3. "
prices = re.findall(r'\$\d+', price_text)

print( prices)
```

```python
import re

medical_record = "Patient A was admitted on Monday. We discussed the treatment with patient B, while Patient X and patient Y remained in the waiting room. Patient alpha was not affected."

redacted_record = re.sub(r'[Pp]atient\s+[A-Z]\b', '[REDACTED]', medical_record)

print("Original Record:\n", medical_record)
print("\nRedacted Record:\n", redacted_record)
```
